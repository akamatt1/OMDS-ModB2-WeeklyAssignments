# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import numpy as np

# Data (approximate values based on Figure 8.10/8.19)
years = np.array([2008, 2009, 2010, 2011, 2012, 2013, 2014])
product_a = [395, 420, 410, 390, 300, 270, 260]
product_b = [360, 400, 390, 360, 290, 260, 250]
product_c = [np.nan, np.nan, 210, 240, 230, 200, 180] # Launched 2010
product_d = [np.nan, np.nan, np.nan, 190, 220, 215, 210] # Launched 2011
product_e = [np.nan, np.nan, np.nan, np.nan, np.nan, 160, 215] # Launched 2013

fig, ax = plt.subplots(figsize=(10, 6))

# Colors (SWD style: Greyscale for context, Blue for recommendation)
context_color = '#A6A6A6' # Grey
highlight_color = '#1F78B4' # Blue

# Plot existing products in grey
ax.plot(years, product_a, color=context_color, linewidth=2)
ax.plot(years, product_b, color=context_color, linewidth=2)
ax.plot(years, product_c, color=context_color, linewidth=2)
ax.plot(years, product_d, color=context_color, linewidth=2)
ax.plot(years, product_e, color=context_color, linewidth=2)

# Label lines directly (Lesson 3: Eliminate Clutter)
ax.text(2014.1, 260, 'A', color=context_color, fontweight='bold')
ax.text(2014.1, 250, 'B', color=context_color, fontweight='bold')
ax.text(2014.1, 180, 'C', color=context_color, fontweight='bold')
ax.text(2014.1, 210, 'D', color=context_color, fontweight='bold')
ax.text(2014.1, 215, 'E', color=context_color, fontweight='bold')

# Recommendation Range (The "Story" - Lesson 6)
# Shaded region between $150 and $200
ax.fill_between([2014.1, 2014.6], 150, 200, color=highlight_color, alpha=0.3)
ax.text(2014.8, 175, 'Recommended\nrange\n$150 - $200', color=highlight_color, fontweight='bold', va='center')

# Average Marker
ax.scatter(2014, 223, color='black', zorder=5)
ax.text(2013.8, 223, 'AVG', ha='right', va='center', fontweight='bold')

# Formatting (Lesson 5: Think like a Designer)
ax.set_title('To be competitive, we recommend introducing our product below\nthe $223 average price point in the $150–$200 range', 
             loc='left', fontsize=14, pad=20)
ax.set_ylabel('Average price', loc='top', rotation=0, color=context_color)
ax.set_xlabel('Year', loc='left', color=context_color)

# Clean up axes
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color(context_color)
ax.spines['bottom'].set_color(context_color)
ax.set_ylim(0, 500)
ax.set_xlim(2008, 2015.5)
ax.set_xticks(years)
ax.tick_params(axis='both', colors=context_color)

plt.tight_layout()
plt.show()
```

Why this reproduction captures the "Pulling it All Together" spirit:
Action-Oriented Title: Instead of "Price over Time," the title is a recommendation. This is the "What" of the Big Idea identified in Lesson 1.

Decluttering (Lesson 3): No gridlines, no border, and the legend is removed in favor of direct labeling. This reduces the cognitive load on the audience.

Preattentive Attributes (Lesson 4): Most of the chart is grey (pushed to the background), while the blue shaded area and text pull your eyes directly to the recommended price range.

Alignment (Lesson 5): The title and labels are aligned to the upper-left, creating a "clean" starting point for the reader's eyes.
