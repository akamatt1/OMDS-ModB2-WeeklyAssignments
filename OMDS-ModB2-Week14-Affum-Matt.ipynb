Here are some types of analysis you can do:

* Run t-tests and Mann Whitney U tests to compare different sets of numbers.
  * For example, if you have a dataset of blood pressure results from hospital patients, you could compare men's results to women's results.
* Run ANOVA tests to compare more than two sets of numbers.
  * For example, you could compare blood pressure results from four different age groups.
* Run chi-squared tests to compare different sets of numbers with categorical variables for rows and columns.
  * For example, if you have the number of patients for men and women, divided according to four age groups, you could find out if these two categorical variables are statistically independent (the number of men in each category approximately equals some fixed number times the number of women).

Working on your datasets: Medical Insurance AnalysisUsing the medical_insurance.csv dataset

I performed several statistical tests to investigate relationships between patient demographics and their medical costs.

1. Analysis Results:
T-test & Mann-Whitney U (Gender vs. Annual Cost)
- Question: Do males and females have different average annual medical costs?
- T-test p-value: 0.3195
- Mann-Whitney U p-value: 0.1347
- Conclusion: In both cases, the p-value is greater than 0.05. We fail to reject the null hypothesis, meaning there is no statistically significant difference in medical costs between sexes in this dataset.

2. ANOVA (Region vs. Annual Cost)
- Question: Does the geographic region (North, South, East, West, Central) affect the average annual medical cost?
- ANOVA p-value: 0.1102
- Conclusion: The p-value is > 0.05. Regional differences in medical costs are not statistically significant.

3. Chi-Squared Test (Sex vs. Smoking Status)
- Question: Is smoking status (Never, Former, Current) independent of a person's sex?
- Chi-squared p-value: 0.1376
- Conclusion: Because $p > 0.05$, we conclude that smoking habits and sex are statistically independent in this population.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 228-255 (chapter nine, Case Study 4 through the end of chapter 10) of the Storytelling With Data book as best you can. You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import numpy as np

# Data from Figure 9.32: Slopegraph
categories = ['Excited', 'Kind of interested', 'OK', 'Bored', 'Not great']
before = [19, 25, 40, 11, 5]
after = [38, 30, 14, 12, 6]

# Define colors: Blue for positive categories, grey for others
colors = ['#1F78B4', '#A6CEE3', '#D3D3D3', '#D3D3D3', '#D3D3D3']

fig, ax = plt.subplots(figsize=(8, 8))

# Plot lines and points
for i in range(len(categories)):
    # Plot line
    color = colors[i]
    linewidth = 3 if i < 2 else 1.5
    ax.plot([0, 1], [before[i], after[i]], color=color, linewidth=linewidth, marker='o')

    # Add text for values on the left (Before)
    ax.text(-0.05, before[i], f"{before[i]}%", color=color, va='center', ha='right', fontsize=12, fontweight='bold' if i < 2 else 'normal')

    # Add text for values on the right (After)
    ax.text(1.05, after[i], f"{after[i]}%", color=color, va='center', ha='left', fontsize=12, fontweight='bold' if i < 2 else 'normal')

    # Add category labels on the right
    ax.text(1.2, after[i], categories[i], color=color, va='center', ha='left', fontsize=12, fontweight='bold' if i < 2 else 'normal')

# Formatting
ax.set_xlim(-0.5, 2.0)
ax.set_ylim(0, 50)

# Titles and Headers
ax.text(0, 48, 'BEFORE', fontsize=14, fontweight='bold', ha='center', color='#555555')
ax.text(1, 48, 'AFTER', fontsize=14, fontweight='bold', ha='center', color='#555555')

# Main Titles (SWD Style: Left Aligned)
plt.text(-0.4, 58, 'Pilot program was a success', fontsize=20, fontweight='bold', color='black')
plt.text(-0.4, 55, 'How do you feel about science?', fontsize=16, color='#555555')
plt.text(-0.4, 52, 'Based on survey of 100 students conducted before and after pilot program.', fontsize=10, color='grey', style='italic')

# Story Annotations
ax.text(-0.4, 30, 'BEFORE program, the\nmajority of children felt\njust OK about science.', color='#888888', fontsize=11, fontweight='bold')
ax.text(1.3, 35, 'AFTER program,\nmore children were\nKind of interested &\nExcited about science.', color='#1F78B4', fontsize=11, fontweight='bold')

# Remove all axes
ax.axis('off')

plt.tight_layout()
plt.savefig('slopegraph_reproduction.png')
```

For this reproduction, I have chosen to recreate the Slopegraph from Figure 9.32 (p. 239).

This chart is a powerful alternative to the pie charts shown in Case Study 5. It effectively tells the story of a successful pilot program by showing the direct shift in student sentiment toward science.

Key Design Choices in this Reproduction:
Focus on Change: The slope of the lines immediately shows the audience the direction of the change. The steep upward slope for "Excited" and the sharp downward slope for "OK" highlight the program's success.

Color Strategy: I used a dark blue for the "Excited" and "Kind of interested" categories to draw the eye to the positive results, while using a neutral grey for the other categories to provide context without competing for attention.

Decluttered Workspace: All traditional chart elements—like axes, tick marks, and borders—have been removed to focus entirely on the data points and the narrative text.

Direct Labeling: Labels are placed next to the data points they describe, eliminating the need for a separate legend and reducing the "mental tax" on the reader.

Strategic Annotations: I included the specific callouts from the book to ensure the "Story" is clear: the decline of the "OK" sentiment and the rise of "Excited" students.
