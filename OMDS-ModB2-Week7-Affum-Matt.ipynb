```python
import numpy as np
import seaborn as sns
import pandas as pd
```

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 99-119 of the Storytelling With Data book as best you can. (The first half of chapter four). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt

# 1. Data Reconstruction (Comparison of two points in time)
categories = ['Model A', 'Model B', 'Model C', 'Model D']
pre_values = [15, 35, 42, 20]
post_values = [28, 38, 30, 18]

plt.figure(figsize=(8, 6))

# 2. Create Slopegraph
x = [0, 1]
for i, category in enumerate(categories):
    # Highlight positive change in blue, negative/neutral in gray
    is_improving = post_values[i] > pre_values[i] and category == 'Model A'
    color = '#1c4e80' if is_improving else '#999999'
    linewidth = 3 if is_improving else 1.5
    
    # Plot lines and points
    plt.plot(x, [pre_values[i], post_values[i]], marker='o', color=color, linewidth=linewidth)
    
    # Left & Right Labels (Direct Labeling)
    plt.text(-0.05, pre_values[i], f'{category} {pre_values[i]}%', ha='right', va='center', color=color)
    plt.text(1.05, post_values[i], f'{post_values[i]}%', ha='left', va='center', color=color)

# 3. SWD Style: Remove "Chart Junk"
ax = plt.gca()
ax.set_xlim(-0.5, 1.5)
ax.set_ylim(0, 60)
ax.axis('off')

# 4. Narrative Titles (Chapter 7 Style)
plt.title('Model A shows the most significant improvement', 
          fontsize=16, loc='left', pad=40, fontweight='bold', color='#333333')
plt.suptitle('Performance comparison: Pre-update vs. Post-update', 
             x=0.125, y=0.92, ha='left', fontsize=12, color='#666666')

# 5. Timeline Context
plt.text(0, -5, 'PRE', ha='center', fontsize=12, fontweight='bold', color='#666666')
plt.text(1, -5, 'POST', ha='center', fontsize=12, fontweight='bold', color='#666666')

plt.tight_layout()
```

The Storytelling Strategy (Chapter 7 Principles)
In Chapter 7, the focus shifts from just "making a graph" to "telling a story." This reproduction uses the following techniques:

The Narrative Title: Instead of "Model Performance," the title is a full sentence that tells you exactly what to see: "Model A shows the most significant improvement."

Horizontal Logic: By using a slopegraph, we create a natural left-to-right flow that mimics a timeline ("Pre" to "Post"), which is easier for the human brain to process as a "story of change."

Color as a Storyteller: Only the category showing significant improvement (Model A) is highlighted in a bold blue. The categories that stayed flat or declined (Models B, C, and D) are pushed into the background with a neutral gray.

The "So What?": By labeling the "Before" and "After" clearly at the bottom, we provide the audience with the necessary context to understand the transformation immediately.
